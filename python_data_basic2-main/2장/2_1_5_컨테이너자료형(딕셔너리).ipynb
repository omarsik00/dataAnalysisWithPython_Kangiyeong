```python
# 딕셔너리 생성 ~ 키-값
my_dict = {
    "name":"Harry",
    "age": 27,
    "height": 190,
    "weight" : 99.9
}

my_dict
```

```python
# 아이템 선택/추가
my_dict.keys()
```

```python
my_dict.values()
```

```python
my_dict['age']
```

```python
my_dict['age'] = 28 #키를 지정해, 기존 아이템 값 변경
my_dict
```

```python
my_dict.update({'weight': 100})#기존 값 없데이트
my_dict
```

```python
my_dict.update({'address': 'Busan'}) #새로운 쌍 추가
my_dict
```
